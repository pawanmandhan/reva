```python
# Here is the **fully updated complete code** including the transformed PCA features section:

# Print many statements at same time using the below command
from IPython.core.interactiveshell import InteractiveShell
InteractiveShell.ast_node_interactivity = "all"

# Import the dataset
import pandas as pd
import numpy as np

wine = pd.read_csv('https://archive.ics.uci.edu/ml/'
                   'machine-learning-databases/wine/wine.data',
                   header=None)

wine.columns = ['Class_label', 'Alcohol', 'Malic_acid', 'Ash',
                'Alcalinity_of_ash', 'Magnesium', 'Total_phenols',
                'Flavanoids', 'Nonflavanoid_phenols', 'Proanthocyanins',
                'Color_intensity', 'Hue',
                'OD280/OD315_of_diluted_wines', 'Proline']

# Data Exploration
wine.head()
wine.shape
wine.Class_label.nunique()
wine.groupby('Class_label').size()
wine.columns

means = np.round(pd.pivot_table(
    wine,
    values=['Alcohol', 'Malic_acid', 'Ash', 'Alcalinity_of_ash',
            'Magnesium', 'Total_phenols', 'Flavanoids',
            'Nonflavanoid_phenols', 'Proanthocyanins',
            'Color_intensity', 'Hue',
            'OD280/OD315_of_diluted_wines', 'Proline'],
    index='Class_label',
    aggfunc=np.mean), 2)

means


# Create Features and Labels
X = wine.drop('Class_label', axis=1)
y = wine['Class_label']

X.head()
y.head()


# Standardize the data
from sklearn.preprocessing import scale

X_scaled = scale(X)
X_scaled = pd.DataFrame(X_scaled, columns=X.columns)

X_scaled.head()
X_scaled.describe()


# Fit PCA model
from sklearn.decomposition import PCA

pca = PCA().fit(X_scaled)

# Explained variance ratio
str(pca.explained_variance_ratio_)
str(pca.explained_variance_ratio_.sum())

# Fit PCA with 5 components
pca = PCA(5).fit(X_scaled)
str(pca.explained_variance_ratio_.sum())

# Kaiser Criteria (Eigenvalues)
pca.explained_variance_

# Create dataframe of components (loadings)
pca_df = pd.DataFrame(pca.components_, columns=X.columns)
pca_df


# Heatmap of components
import matplotlib.pyplot as plt
%matplotlib inline
plt.figure(figsize=(10, 10))

import seaborn as sns
sns.heatmap(pca_df, cmap='RdYlGn', annot=True)

plt.show()


# Original features
X_scaled.shape   # (178, 13)

# Transform the data into principal components
X_scaled_trnsf = pca.fit_transform(X_scaled)

# Transformed features (Now 5 components instead of 13)
X_scaled_trnsf = pd.DataFrame(X_scaled_trnsf,
                              columns=('P1', 'P2', 'P3', 'P4', 'P5'))

X_scaled_trnsf.head()
X_scaled_trnsf.shape

```

```python

# ============================================================
# Relationship Between Principal Components
# ============================================================

# IMPORTANT:
# PCA components are orthogonal (at 90 degrees to each other).
# Orthogonality ⇒ Covariance = 0 ⇒ Correlation ≈ 0
# Let's verify this mathematically and visually.


# 1️⃣ Correlation Matrix of Principal Components
# ------------------------------------------------
# If PCA worked correctly, off-diagonal values should be ~0

pc_corr = X_scaled_trnsf.corr()
pc_corr


# 2️⃣ Heatmap of PCA Component Correlation
# ------------------------------------------------
# Diagonal values = 1 (self-correlation)
# Off-diagonal values ≈ 0 (components are uncorrelated)

plt.figure(figsize=(6, 5))
sns.heatmap(pc_corr, annot=True, cmap='coolwarm', center=0)
plt.title("Correlation Between Principal Components")
plt.show()


# 3️⃣ Direct Correlation: P1 vs P2
# ------------------------------------------------
# Using numpy to directly compute correlation
# Expectation: Very close to 0 (may see tiny values like 1e-16 due to floating point precision)

np.corrcoef(X_scaled_trnsf['P1'], X_scaled_trnsf['P2'])


# 4️⃣ Scatter Plot: P1 vs P2
# ------------------------------------------------
# If components are independent (uncorrelated),
# there should be no linear trend between them.

plt.figure(figsize=(6, 5))
plt.scatter(X_scaled_trnsf['P1'], X_scaled_trnsf['P2'])
plt.xlabel('P1')
plt.ylabel('P2')
plt.title('P1 vs P2 (Orthogonal Principal Components)')
plt.axhline(0)
plt.axvline(0)
plt.show()


# ============================================================
# INTERPRETATION:
# ------------------------------------------------------------
# ✔ PCA components are eigenvectors of the covariance matrix.
# ✔ Eigenvectors are orthogonal to each other.
# ✔ Orthogonality ⇒ Zero covariance ⇒ Zero correlation.
# ✔ This is how PCA removes multicollinearity.
# ============================================================

### 🎯 Interview-Ready One-Liner

# PCA components are orthogonal eigenvectors of the covariance matrix, so they are linearly uncorrelated — correlation between PC1 and PC2 is theoretically zero.


# If you'd like, I can also add:

# * Scree plot 📊
# * Cumulative explained variance curve
# * 2D PCA projection colored by wine class
# * Mathematical intuition diagram
```
